def cost_function(X, Y, W):
    """ Parameters:
This function finds the Mean Square Error.
Input parameters:
X: Feature Matrix
Y: Target Matrix
W: Weight Matrix
Output Parameters:
cost: accumulated mean square error."""
    xt=np.transpose(X)
    ypred=np.dot(W,xt)
    loss=(Y-ypred)**2
    cost=np.sum(loss)/X.shape[0]
    return cost

```python
def gradient_descent(X, Y, W, alpha, iterations):
    """
Perform gradient descent to optimize the parameters of a linear regression model.
Parameters:
X (numpy.ndarray): Feature matrix (m x n).
Y (numpy.ndarray): Target vector (m x 1).
W (numpy.ndarray): Initial guess for parameters (n x 1).
alpha (float): Learning rate.
iterations (int): Number of iterations for gradient descent.
Returns:
tuple: A tuple containing the final optimized parameters (W_update) and the history of cost values
.
W_update (numpy.ndarray): Updated parameters (n x 1).
cost_history (list): History of cost values over iterations.
"""
    # Initialize cost history
    cost_history = [0] * iterations
    # Number of samples
    m = len(Y)
    for iteration in range(iterations):
        # Step 1: Hypothesis Values
        xt=np.transpose(X)
        Y_pred = np.dot(W,xt)
        # Step 2: Difference between Hypothesis and Actual Y
        loss = (Y-Y_pred)**2
        # Step 3: Gradient Calculation
        dw = np.sum(np.dot(loss,X))/m
        # Step 4: Updating Values of W using Gradient
        W_update = W-alpha*dw
        W=W_update
        # Step 5: New Cost Value
        cost = cost_function(X, Y, W_update)
        cost_history[iteration] = cost
        return W_update, cost_history
```

```python
# Model Evaluation - RMSE
def rmse(Y, Y_pred):
    """
This Function calculates the Root Mean Squres.
Input Arguments:
Y: Array of actual(Target) Dependent Varaibles.
Y_pred: Array of predeicted Dependent Varaibles.
Output Arguments:
rmse: Root Mean Square.
"""
    n= len(Y)
    dw = np.sum((Y-Y_pred)**2)/n
    rmse = np.sqrt(dw)
    return rmse
```

```python
import numpy as np
import pandas as pd
def train_test_split_scratch(X, y, test_size=0.2, random_seed=42):
    """
    Custom train-test split without sklearn.
    """
    np.random.seed(random_seed)
    indices = np.arange(X.shape[0])
    np.random.shuffle(indices)
    test_size_count = int(len(X) * test_size)
    test_idx = indices[:test_size_count]
    train_idx = indices[test_size_count:]
    return X[train_idx], X[test_idx], y[train_idx], y[test_idx]

```
